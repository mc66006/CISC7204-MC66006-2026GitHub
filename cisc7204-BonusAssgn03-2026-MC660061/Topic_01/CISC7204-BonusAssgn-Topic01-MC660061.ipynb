### Pandas Series Part 2

- Using comparison operators to produce Boolean series
- Using Boolean indexing to filter data
- Using OR and AND operations to make compound filters
- Assigning subsets to their own variable
- Operating on subsets in place using `.loc`

```python
import pandas as pd
```

```python
# Base Python uses numeric indexes to return a single element
ser = range(-2, 3)
ser[0]
```

```python
# Let's make a pandas series to go into pandas indexing
ser = pd.Series(ser)
ser
```

```python
# The corresponding index for the first element is True and the rest are False
# Using the corresponding index of a Boolean collection to filter is called "Boolean masking"
first = [True, False, False, False, False]
ser[first]
```

```python
# If all Boolean values are True, we return the original series
all_true = [True, True, True, True, True]
ser[all_true]
```

```python
all_false = [False, False, False, False, False]
ser[all_false]
```

```python
# The Boolean mask is filtering our results here
first_and_third = [True, False, True, False, False]
ser[first_and_third]
```

```python
# Boolean masking leaves the original series intact
ser
```

```python
# Comparison operators return a Boolean series
# Notice this returns a Boolean series
ser == 1
```

```python
mask = ser == 1
ser[mask]
```

```python
# We can place the Boolean series inside the square brackets directly
ser[ser == 1]
```

```python
# Using a Boolean series in square brackets after the series filters results
is_negative = ser < 0
is_negative
```

```python
# The True values in the Boolean series enable the corresponding elements. False values hide corresponding elements.
ser[is_negative]
```

```python
# Subsets are copies of the data
negatives = ser[is_negative]
negatives
```

```python
# Reassigning the result of a Boolean mask keeps the original series intact
ser
```

```python
# A comparison operator used with the modulo operator to return a Boolean Series
is_odd = ser % 2 == 1
ser[is_odd]
```

```python
# Let's work with some new data
numbers = pd.Series(range(1, 13))
numbers
```

```python
# We can use & and | operators on our Boolean series to produce more complex behavior
# Parentheses are helpful for order of operations
numbers[(numbers == 2) | (numbers == 5)]
```

```python
# If all expressions evaluate to false, we can an empty series
numbers[(numbers == 2) & (numbers == 5)]
```

```python
# To avoid parentheses, we can assign each Boolean series separately. 
# The & operator is the AND operator for Boolean series
is_even = numbers % 2 == 0
is_divisible_by_3 = numbers % 3 == 0
is_divisible_by_3_or_2 = is_even & is_divisible_by_3
numbers[is_divisible_by_3_or_2]
```

```python
# If each Boolean series/expression is not assigned to its own variable, we need parentheses for order of operations
# Notice how with an AND operator, both Booleans must be true
numbers[(numbers % 2 == 0) & (numbers % 5 == 0)]
```

```python
# We use the | operator for OR operations 
# This returns all the even numbers or the numbers evenly divisible by 5
numbers[(numbers % 2 == 0) | (numbers % 5 == 0)]
```

```python
# Boolean masking is very powerful, but what about modifying values in place on a series?
# The .loc method use the same Boolean series syntax
is_even = numbers % 2 == 0

# For simplicity, let's assign every even number to 200
numbers.loc[is_even] = 200
```

```python
numbers
```

```python
# What if we need more dynamic reassignment?
numbers = pd.Series(range(1, 13))

# Shorthand syntax would be numbers.loc[is_even] *= 2
numbers.loc[is_even] = numbers.loc[is_even] * 2
numbers
```

- Make a series of the following values `[-4, -3, -2, -1, 0, 1, 2, 3, 4]` and store to a variable named `ser`
- Write the code to filter out only the number 2
- Write the code to filter out the number 2 or 4.
- Create a variable named `is_positive` that will be the Boolean series if the corresponding `ser` values is positive or not. Now create a new variable named `positives` and use this to store only the positive numbers.
- Follow the steps above creating a Boolean series named `is_even` and assign a new variable named `evens` that holds only the even numbers
- Use what you have learned to produce a new Boolean series named `is_even_and_positive` and assign the Boolean values. Use your new Boolean series to create a new variable named `even_positives` that is only the even and positive numbers.
- Follow the pattern from above to produce a new Boolean series named `is_even_or_positive` and assign the appropriate Boolean values using | for OR operations. Use your new Boolean series to create a new variable named `even_or_positive` that is only the even and positive numbers
- Use the `.loc` method to reassign all numbers that are both even and positive to be the number zero.
- Use the `.loc` method and the reassignment syntax to multiply every negative number by 20, in place.

```python
# Make a series out of  [-4, -3, -2, -1, 0, 1, 2, 3, 4]


ser = pd.Series([-4, -3, -2, -1, 0, 1, 2, 3, 4])
ser
```

```python
# Write the code to filter out only the number 2


ser[ser == 2]
```

```python
# Write the code to filter out the number 2 or 4.


ser[(ser == 2) | (ser == 4)]
```

```python
# Create a variable named `is_positive` that will be the Boolean series if the corresponding `ser` values is positive or not. Now create a new variable named `positives` and use this to store only the positive numbers.


is_positive = ser > 0
positives = ser[is_positive]
positives
```

```python
# Follow the steps above creating a Boolean series named `is_even` and assign a new variable named `evens` that holds only the even numbers


is_even = ser % 2 == 0
evens = ser[is_even]
evens
```

```python
# Use what you have learned to produce a new Boolean series named `is_even_and_positive` and assign the Boolean values. Use your new Boolean series to create a new variable named `even_positives` that is only the even and positive numbers.
is_even_and_positive = is_even & is_positive
even_positives = ser[is_even_and_positive]
even_positives
```

```python
# Follow the pattern from above to produce a new Boolean series named `is_even_or_positive` and assign the appropriate Boolean values using | for OR operations. Use your new Boolean series to create a new variable named `even_or_positive` that is only the even and positive numbers


is_even_or_positive = is_even | is_positive
even_or_positive = ser[is_even_or_positive]
even_or_positive
```

```python
# Use the `.loc` method to reassign all numbers that are both even and positive to be the number zero.


ser.loc[is_even_and_positive] = 0
ser
```

```python
# Use the `.loc` method and the reassignment syntax to multiply every negative number by 20, in place.

is_negative = ser < 0
ser.loc[is_negative] = ser.loc[is_negative] * 20
ser
```

### Pandas Series Part 3: Strings
- Sorting values
- Using pandas built-in string methods
- Assigning and reassigining results
- Using string methods for data cleaning
- Updating data types

```python
import pandas as pd
```

```python
fruits = pd.Series(["apple", "orange", "banana", "lemon", "lime", "pineapple", "blueberry", "raspberry", "cranberry"])
fruits
```

```python
# .sort_values sorts strings alphabetically or numbers in numerical order
# fruits.sort_values(ascending=True) the default sort order
fruits.sort_values()
```

```python
fruits.sort_values(ascending=False)
```

```python
# Use inplace=True to operate on the original
# For more on .sort_values, see https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html
fruits = fruits.sort_values()
fruits
```

```python
# We can reassign the series to hold the sorted values
# For more on .sort_values, see https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html
fruits = fruits.sort_values(ignore_index=True)
fruits
```

```python
# .capitalize to capitalize
fruits.str.capitalize()
```

```python
fruits
```

```python
# String operations keep the original series intact, so we reassign to update
capitalized_fruits = fruits.str.capitalize()
capitalized_fruits
```

```python
# .contains returns a boolean series
# Always remember to use .str or your results will be in error
fruits.str.contains("apple")
```

```python
# Since .contains returns a Boolean series, we can use it to filter our results
fruits[fruits.str.contains("apple")]
```

```python
# .count to count substring occurrences
fruits.str.count("a")
```

```python
fruits.str.count("berry")
```

```python
# Summing up the results of .count
vowel_counts = fruits.str.count("a") + fruits.str.count("e") + fruits.str.count("i") + fruits.str.count("o") + fruits.str.count("u")
vowel_counts
```

```python
# Using count with a Regular Expression character class
# Some of the Pandas string methods can utilize regular expressions
fruits.str.count("[aeiou]")
```

```python
# We can use our new vowel count to filter values from the series
fruits[fruits.str.count("[aeiou]") > 2]
```

```python
# .startswith returns a Boolean series
fruits.str.startswith("l")
```

```python
fruits[fruits.str.startswith("l")]
```

```python
# .endswith returns a Boolean series
fruits.str.endswith("berry")
```

```python
# .len to get the length of the string
fruits.str.len()
```

```python
# .lower to lowercase strings
shouts = pd.Series(["PLEASE", "LOWERCASE", "THESE", "STRINGS"])
not_shouts = shouts.str.lower()
not_shouts
```

```python
# Using .replace to replace characters (also used to remove characters)
prices = pd.Series(["€5.99", "€12.25", "€95"])

# Be sure to reassign the variable
prices = prices.str.replace("€", "")

# But our data type is still a string
prices * 2
```

```python
# Use .astype to convert a number in a string to a numeric data type
prices = prices.astype(float)
prices * 2
```

```python
# .upper to uppercase
fruits.str.upper()
```

- Now make a new series named `prices` that holds `["$2.99", "$1,200.25", "$5.99", "$2,350.00"]`
- Reassign `prices` to hold only a string of numbers. Remove the `$` and `,` characters.

- Now multiply your `prices` series by `0.9`

```python
# Create a series of vegetables ["Onion", "cucumber", "Carrot", "squash", "Potato", "Asparagus", "kale", "Broccoli", "spinach"]


vegetables = pd.Series([
    "Onion", "cucumber", "Carrot", "squash", "Potato",
    "Asparagus", "kale", "Broccoli", "spinach"
])
vegetables
```

```python
# Write the code necessary to lowercase all of the vegetables and reassign your series.


vegetables = vegetables.str.lower()
vegetables
```

```python
# Write the pandas code to sort the strings in alphabetical order. Ensure that the series stores the sorted order


vegetables = vegetables.sort_values(ignore_index=True)
vegetables
```

```python
# Write the pandas code to show only the vegetables that start with a vowel


vowels = "aeiou"
starts_with_vowel = vegetables.str.startswith(tuple(vowels))
vegetables[starts_with_vowel]
```

```python
# Write the pandas code to show the vegetables that have exactly two vowels


vowel_count = vegetables.str.count("[aeiou]")
vegetables[vowel_count == 2]
```

```python
# Make a new series named prices that holds ["$2.99", "$1,200.25", "$5.99", "$2,350.00"]


prices = pd.Series(["$2.99", "$1,200.25", "$5.99", "$2,350.00"])
prices
```

```python
# Reassign prices to hold only a string of numbers. Remove the $ and , characters.


prices = prices.str.replace("$", "", regex=False)
prices = prices.str.replace(",", "", regex=False)
prices
```

```python
# Reassign prices to be a float data type


prices = prices.astype(float)
prices
```

```python
# Multiply your prices series by 0.9


prices * 0.9
```
